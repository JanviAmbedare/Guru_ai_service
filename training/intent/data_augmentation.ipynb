```python
import random
import pandas as pd

# ======================================================
# ADVANCED AUGMENTATION RULES
# ======================================================

PREFIXES = [
    "",
    "please",
    "guru",
    "hey guru",
    "can you",
    "could you",
    "would you",
    "guru please",
    "hey",
    "listen guru"
]

SUFFIXES = [
    "",
    "please",
    "right now",
    "immediately",
    "for me",
    "quickly"
]

HINGLISH_VARIATIONS = {

    "MOVE_FORWARD":[
        "aage chalo",
        "aage jao"
    ],

    "STOP":[
        "ruk jao",
        "bas karo"
    ],

    "TURN_LEFT":[
        "baaye jao"
    ],

    "TURN_RIGHT":[
        "daaye jao"
    ],

    "TAKE_PHOTO":[
        "photo lo"
    ],

    "CAMERA_ON":[
        "camera chalu karo"
    ],

    "FIND_OBJECT":[
        "meri bottle dhoondo"
    ]
}

TYPO_VARIATIONS = {

    "battery":[
        "battry"
    ],

    "camera":[
        "camra"
    ],

    "reminder":[
        "remnder"
    ]
}

```

```python

def clean_text(text):

    return (
        text.lower()
        .strip()
    )


# TYPO GENERATOR

def add_typo_variation(text):

    words = text.split()

    updated = []

    for word in words:

        if (
            word in TYPO_VARIATIONS and
            random.random() < 0.3
        ):

            updated.append(
                random.choice(
                    TYPO_VARIATIONS[word]
                )
            )

        else:

            updated.append(word)

    return " ".join(updated)


# HINGLISH AUGMENTATION
def add_hinglish(text):

    if text in HINGLISH_VARIATIONS:

        return random.choice(
            HINGLISH_VARIATIONS[text]
        )

    return text


```

```python
import random
import re


MAX_AUGMENTATIONS = 4


def replace_first(text, old, new):
    """Replace only the first occurrence."""
    return re.sub(rf"\b{re.escape(old)}\b", new, text, count=1)


def augment_sentence(sentence, intent):

    sentence = sentence.strip().lower()

    augmented = {sentence}

    # ------------------------------------------
    # Prefix
    # ------------------------------------------

    if random.random() < 0.60:
        prefix = random.choice(PREFIXES)
        if prefix:
            augmented.add(f"{prefix} {sentence}")

    # ------------------------------------------
    # Suffix
    # ------------------------------------------

    if random.random() < 0.50:
        suffix = random.choice(SUFFIXES)
        if suffix:
            augmented.add(f"{sentence} {suffix}")

    # ------------------------------------------
    # Hinglish
    # ------------------------------------------

    if intent in HINGLISH_VARIATIONS:

        if random.random() < 0.30:

            augmented.update(HINGLISH_VARIATIONS[intent])

    # ------------------------------------------
    # Typos
    # ------------------------------------------

    for word, variants in TYPO_VARIATIONS.items():

        if word in sentence and random.random() < 0.15:

            typo = random.choice(variants)

            augmented.add(
                replace_first(sentence, word, typo)
            )

    # ------------------------------------------
    # Random Sampling
    # ------------------------------------------

    augmented = list(augmented)

    random.shuffle(augmented)

    return augmented[:MAX_AUGMENTATIONS]
```

```python

# DATASET GENERATION

def generate_dataset(base_data):

    rows = []

    for intent, sentences in base_data.items():

        for sentence in sentences:

            variants = augment_sentence(
                sentence,
                intent
            )

            for text in variants:

                rows.append({
                    "text": text,
                    "intent": intent
                })

    return pd.DataFrame(rows)
```
